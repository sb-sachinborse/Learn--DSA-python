### 977. Squaring a sorted array

```python
nums = [-4,-1,0,3,10]
# output = [0,1,9,16,100]


def sortedSquares(nums):
    left = 0
    right = len(nums) -1
    result = []

    while left <= right:
        if abs(nums[left]) > abs(nums[right]):
            result.append(nums[left] ** 2)
            left+=1
        else:
            result.append(nums[right]**2)
            right-=1
    result.reverse()
    return result


# approch 2: 
# def sortedSquares(nums):
#     n = len(nums)
#     left = 0
#     right = n -1
#     k = n-1
#     result = [0] * n

#     while left <= right:
#         if abs(nums[left]) > abs(nums[right]):
#             result[k] = nums[left] * nums[left]
#             left+=1
#         else:
#             result[k] = nums[right] * nums[right] 
#             right-=1
#         k -= 1
#     return result



sortedSquares(nums)
```

### 125. Valid Palindrome

```python
s = "A man , a plane, a canal: Panama"   
#s= "race a car"

def isValidPlindrome(s)-> bool:
    
    i,j = 0,len(s) -1

    def isAlphanumeric(ch):
        code = ord(ch)
        return (48<= code <=57) or (65<= code <= 90) or (97 <= code <= 122)
    
    while i < j:
        left ,right = s[i],s[j]
        
        if not isAlphanumeric(left):
            i += 1
            continue
        if not isAlphanumeric(right):
            j -= 1
            continue

        if left.lower() != right.lower():
            return False
        
        i+=1
        j-=1
        return True


isValidPlindrome(s)
    
```

### 344. Reverse String

```python
s = ["h","e","l","l","o"]
#output = ["o","l","l","e","h"]

def reverString(s) :
    left = 0
    right = len(s) - 1

    while left < right:
        s[left],s[right] = s[right],s[left]
        left += 1
        right -= 1
    return s

reverString(s)

#Time:O(n) linear
#Space:O(1) constant
```

### 680 .Valid Palindrome 2

```python
#we can delete one char  only
s = "abxa"

def validPalindrome(s: str) -> bool:
    
    def is_palindrome(l, r):
        while l < r:
            if s[l] != s[r]:
                return False
            l += 1
            r -= 1
        return True

    left, right = 0, len(s) - 1

    while left < right:
        if s[left] != s[right]:
            # try skipping either left or right character
            return is_palindrome(left + 1, right) or is_palindrome(left, right - 1)
        left += 1
        right -= 1

    return True

validPalindrome(s)

#Time:O(n) linear
#space :O(1) constat
```

### valid Word Abbrevation    (neetcode.io)

```python
word = "apple"
abbr = "a3e"


def validWordAbbrivation(word,abbr)-> bool:

    i = 0
    j = 0
 

    while i < len(word) and j < len(abbr):
        #case1.digit in abbreavation
        if abbr[j].isdigit():
            #leading zero is valid
            if abbr[j] == '0':
                return False
            
            num = 0
            while j<len(abbr) and abbr[j].isdigit():
                num = num * 10 + int(abbr[j])
                j+=1
            i+=num  #skip char in word

        #case2:letter in abbreavation
        else:
            if i>= len(word) or word[i]  != abbr[j]:
                return False

            i+=1
            j+=1
    return i == len(word) and j==len(abbr)


validWordAbbrivation(word,abbr)

#Time:O(m+n)
#Space:O(1)
```

### 88. Merge Sorted Array

```python
nums1=[1,2,3,0,0,0] 
m=3
nums2=[2,5,6] 
n=3


def merge(nums1, m, nums2, n):
    i = m - 1   # last element in nums1
    j = n - 1   # last element in nums2
    k = m + n - 1  # last position in nums1

    while j >= 0:
        if i >= 0 and nums1[i] > nums2[j]:
            nums1[k] = nums1[i]
            i -= 1
        else:
            nums1[k] = nums2[j]
            j -= 1
        k -= 1

    


merge(nums1,m,nums2,n)
```
