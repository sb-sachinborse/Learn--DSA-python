#### Squaring a sorted array

```python
nums = [-4,-1,0,3,10]
# output = [0,1,9,16,100]


def sortedSquares(nums):
    left = 0
    right = len(nums) -1
    result = []

    while left <= right:
        if abs(nums[left]) > abs(nums[right]):
            result.append(nums[left] ** 2)
            left+=1
        else:
            result.append(nums[right]**2)
            right-=1
        result.reverse()
        return result

print(sortedSquares(nums))
```

### 9. Palindrome Number

```python
x = 1214


def isPlindrome(x:int) -> bool:
    
    if str(x) == str(x)[::-1]:
        return True
    else:
        return False


isPlindrome(x)
```

### 125. Valid Palindrome

```python
s = "A man , a plane, a canal: Panama"   
#s= "race a car"

def isValidPlindrome(s)-> bool:
    
    i,j = 0,len(s) -1

    def isAlphanumeric(ch):
        code = ord(ch)
        return (48<= code <=57) or (65<= code <= 90) or (97 <= code <= 122)
    
    while i < j:
        left ,right = s[i],s[j]
        
        if not isAlphanumeric(left):
            i += 1
            continue
        if not isAlphanumeric(right):
            j -= 1
            continue

        if left.lower() != right.lower():
            return False
        
        i+=1
        j-=1
        return True


isValidPlindrome(s)
    
```
